# Tables

## Check Table Exist

```python
cursor = conn.cursor()

try:
    cursor.execute(
        "SELECT EXISTS (SELECT 1 FROM information_schema.tables WHERE table_name = %s)", ('countries',)
    )
    tableExist = cursor.fetchone()[0]
    print(tableExist)
except Exception as e:
    print(f"An error occurred: {e}")
    conn.rollback()
finally:
    cursor.close()
```

## Drop table if it exist.

```python
cursor = conn.cursor()

try:
    table_name = 'countries'
    cursor.execute(f"DROP TABLE {table_name}")
    print("Table dropped!")
    # Commit your changes in the database
    conn.commit()
except Exception as e:
    print(f"An error occurred: {e}")
    conn.rollback()
finally:
    cursor.close()
```

## Create Table
### Countries Table

Country names are unique and this table stores the default country names.

- **id** `SERIAL PRIMARY KEY` → Unique identifier for each country  
- **name** `VARCHAR(100) NOT NULL UNIQUE` → The name of the country and not allow null

```python
# Create "countries" table.
cur = conn.cursor()
countryTableCreateQuery = """ 
            CREATE TABLE countries (
                id SERIAL PRIMARY KEY,
                name VARCHAR(100) NOT NULL UNIQUE
            );
            """

try:
    cur.execute(countryTableCreateQuery)
    conn.commit()
except Exception as e:
    print(f"An error occurred: {e}")
    conn.rollback()
finally:
    cur.close()
```

# APIs

### Call API for flask app.

```python
import urllib3
import json

http = urllib3.PoolManager()

try:
    response = http.request('GET', 'http://apan5210-flask-app:5210/api/random-number')

    if response.status == 200:
        data = json.loads(response.data.decode('utf-8'))
        print(f"New API data: {data}")
    else:
        print(f"An error occurred: {response.status}")
except urllib3.exceptions.HTTPError as e:
    print(f"An error occurred: {e}")
```
